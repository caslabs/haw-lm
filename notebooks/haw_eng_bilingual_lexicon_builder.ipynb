````python
data = api.hawaiian_to_english
pretty_json = json.dumps(data, indent=4, ensure_ascii=False)
````

````python
# Here, we will convert the data into a list of records to be used in a DataFrame
records = []
for k, rows in data.items():
    # Concenate tuple of lists into a single list
    for row in rows:
        if isinstance(row, tuple):
            row = list(row)
        elif not isinstance(row, list):
            row = [row]
        records.append([k] + row)

df = pd.DataFrame(records)
df.columns = ["source"] + [f"col{i}" for i in range(df.shape[1] - 1)]
````

````python
# Convert DataFrame to a more suitable format
# Drop source
df = df.drop(columns=["source"])
# Rename col0 to "haw"
df = df.rename(columns={"col0": "haw"})
# Rename col1 to "eng"
df = df.rename(columns={"col1": "eng"})
````
