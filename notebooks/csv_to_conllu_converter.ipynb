```python
"""
csv-to-conllu.py
Convert a simple 4-column POS CSV into valid CoNLL-U.
The rest of the CoNLL-U columns are filled with placeholders.
since that is not in scope of this project.

Input CSV format (per row):
    0  ID      – token index
    1  FORM    – surface form
    2  LEMMA   – lemma
    3  UPOS    – universal POS
Comment / sentence-header rows start with '#'.
"""
import csv
from pathlib import Path

INPUT  = Path("../data/pos-makana/v1/makana_v1.csv")
OUTPUT = Path("makana_v1.conllu")

with INPUT.open(newline="", encoding="utf-8") as f_in, OUTPUT.open("w", encoding="utf-8") as f_out:
    reader   = csv.reader(f_in)
    sent_id  = 1

    for row in reader:
        if not row or row[0].startswith("#"):
            # Blank line between finished sentences (except before the very first)
            if sent_id > 1:
                f_out.write("\n")

            # Write the sentence-level comments if present
            text = row[0].lstrip("# ").strip() if row else ""
            if text:
                f_out.write(f"# sent_id = {sent_id}\n")
                f_out.write(f"# text = {text}\n")

            sent_id += 1
            continue

        idx, form, lemma, upos = [cell.strip() for cell in row[:4]]

        # Produce the 10 CoNLL-U columns
        #  ID  FORM  LEMMA  UPOS  XPOS  FEATS  HEAD  DEPREL  DEPS  MISC
        conllu_cols = [
            # These are derived from the MAKANA spreadsheet.
            idx,              # ID
            form,             # FORM
            lemma,            # LEMMA
            upos,             # UPOS

            # The rest of the columns are placeholders.
            "_",              # XPOS  (not used)
            "_",              # FEATS (not annotated)
            "0",              # HEAD  (attach to virtual root)
            "dep",            # DEPREL (generic relation)
            "_",              # DEPS
            "_"               # MISC
        ]

        f_out.write("\t".join(conllu_cols) + "\n")

print(f"Wrote {OUTPUT}")
```
