```python
# Set your CoNLL-U file path here
conllu_path = '../data/pos-makana/v1/makana_v2.conllu'  # <-- update this variable

import pandas as pd
import matplotlib.pyplot as plt
import re

def parse_conllu(filepath):
    sentences = 0
    words = 0
    pos_tags = []
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line.startswith('# sent_id'):
                sentences += 1
            elif line and not line.startswith('#'):
                parts = line.split('\t')
                if len(parts) > 3 and re.match(r'^\d+$', parts[0]):
                    words += 1
                    pos_tags.append(parts[3])
    return sentences, words, pos_tags

num_sentences, num_words, pos_tags = parse_conllu(conllu_path)

print(f'Total sentences: {num_sentences}')
print(f'Total words: {num_words}')

# POS tag distribution
pos_counts = pd.Series(pos_tags).value_counts()
plt.figure(figsize=(7, 7))
pos_counts.plot.pie(autopct='%1.1f%%', startangle=90, cmap='tab20')
plt.title('POS Tag Distribution')
plt.ylabel('')
plt.show()
```
